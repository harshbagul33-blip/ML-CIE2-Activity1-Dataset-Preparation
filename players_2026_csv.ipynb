```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values[missing_values > 0])
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Values Percentage:")
print(missing_percentage[missing_percentage > 0].round(2))
```

```python
print("Columns Without Missing Values:")
print(df.columns[df.isnull().sum() == 0].tolist())
```

```python
df_clean = df.copy()
```

```python
age_median = df_clean["Age"].median()

df_clean["Age"] = df_clean["Age"].fillna(age_median)

print("Missing Age Values:", df_clean["Age"].isnull().sum())
```

```python
nation_mode = df_clean["Nation"].mode()[0]

df_clean["Nation"] = df_clean["Nation"].fillna(nation_mode)

print("Missing Nation Values:", df_clean["Nation"].isnull().sum())
```

```python
remaining_missing = df_clean.isnull().sum()

print("Remaining Missing Values:")
print(remaining_missing[remaining_missing > 0])
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
df_clean[df_clean.duplicated()]
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape After Removing Duplicates:", df_clean.shape)
print("Remaining Duplicate Rows:", df_clean.duplicated().sum())
```

```python
categorical = df_clean.select_dtypes(include=['object']).columns

for col in categorical:
    if df_clean[col].dtype == 'object':
        df_clean[col] = df_clean[col].astype(str).str.strip()

print("Extra spaces checked and removed.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df_clean["Age"], bins=15, edgecolor="black")

plt.title("Distribution of Players' Age")
plt.xlabel("Age")
plt.ylabel("Number of Players")

plt.show()
```

```python
print("Before Encoding:")
print(df_clean["Nation"].head())
```

```python
df_encoded = pd.get_dummies(
    df_clean,
    columns=["Nation"],
    dtype=int
)

print("After Encoding:")
display(df_encoded.head())
```

```python
print("Before Encoding:", df_clean.shape)
print("After Encoding:", df_encoded.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaling_columns = ["Age"]

df_scaled = df_encoded.copy()

scaler = StandardScaler()

df_scaled[scaling_columns] = scaler.fit_transform(
    df_scaled[scaling_columns]
)

print("Age after Standardisation:")
display(df_scaled[["Age"]].head())
```

```python
print("Age Before Scaling:")
display(df_encoded[["Age"]].head())

print("Age After Scaling:")
display(df_scaled[["Age"]].head())
```

```python
print("Final Dataset Shape:", df_scaled.shape)
```

```python
print("Total Missing Values:", df_scaled.isnull().sum().sum())
```

```python
print("Remaining Duplicate Rows:", df_scaled.duplicated().sum())
```

```python
df_scaled.to_csv("players_2026_prepared.csv", index=False)

print("Prepared dataset saved successfully.")
```
